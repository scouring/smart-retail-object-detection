```python
from ultralytics import YOLO

# Load a pretrained YOLOv8n model (smallest model to start with)
model = YOLO("yolov8n.pt")

# Fine-tune YOLOv8 on a custom dataset
model.train(
    data=dataset_path,       # points to data.yaml
    epochs=50,               # adjust based on dataset size
    imgsz=640,               # image size
    batch=16,                # lower if out-of-memory

    # save training outputs (weights, logs, plots)
    project="/content/drive/MyDrive/smart_retail/runs",
    name="train_yolov8"
)
```

```python
# load the best trained weights
model = YOLO("/content/drive/MyDrive/smart_retail/runs/train_yolov84/weights/best.pt")

# make predictions on validation images
results = model.predict(
    source="/content/drive/MyDrive/smart_retail/data/images/val",
    conf=0.25,
    save=True,
    project = "/content/drive/MyDrive/smart_retail/runs",
    name="predictions"
)
```
